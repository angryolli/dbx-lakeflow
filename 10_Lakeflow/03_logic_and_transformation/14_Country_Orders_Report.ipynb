```python
from pyspark.sql.functions import col

dbutils.widgets.text("country", "Finland")

country = dbutils.widgets.get("country")

df = (
    spark.read.table(
        "workspace.retail_pipeline.customers_orders_silver"
    )
    .filter(col("country") == country)
)

table_name = (
    "workspace.retail_pipeline.orders_"
    + country.lower()
    + "_silver"
)

df.write.mode("overwrite").saveAsTable(table_name)

print(f"Created {table_name}")
```

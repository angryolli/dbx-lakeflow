```python
%sql
CREATE OR REPLACE TABLE
workspace.retail_pipeline.customers_sales_silver AS
SELECT
    c.customer_id,
    c.name AS customer_name,
    c.country,
    o.order_id,
    o.product,
    s.sale_id,
    s.amount AS total_price,
    TO_DATE(s.sale_date) AS sale_date
FROM workspace.retail_pipeline.customers_bronze c
JOIN workspace.retail_pipeline.orders_bronze o
    ON c.customer_id = o.customer_id
JOIN workspace.retail_pipeline.sales_bronze s
    ON o.order_id = s.order_id;
```

```python
table = "workspace.retail_pipeline.customers_sales_silver"

df = spark.read.table(table)

has_duplicates = df.count() > df.dropDuplicates().count()

dbutils.jobs.taskValues.set(
    key="has_duplicates",
    value=str(has_duplicates).lower()
)

print(f"Duplicates found: {has_duplicates}")
```

```python
%sql
CREATE OR REPLACE TABLE
workspace.retail_pipeline.customers_orders_silver AS
SELECT
    c.customer_id,
    c.name AS customer_name,
    c.country,
    o.order_id,
    o.product,
    o.amount,
    o.status
FROM workspace.retail_pipeline.customers_bronze c
JOIN workspace.retail_pipeline.orders_bronze o
    ON c.customer_id = o.customer_id;
```

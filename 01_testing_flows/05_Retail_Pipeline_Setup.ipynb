# Retail pipeline setup

This notebook starts a new retail pipeline. Before the later ingestion and transformation notebooks can run, it does three things:

1. Choose the catalog and schema: `workspace.retail_pipeline`.
2. Insert the starting rows into the `orders` and `sales` tables.
3. Create the `trigger_storage_location` volume, where files that start the pipeline are stored.

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.retail_pipeline;

USE CATALOG workspace;
USE SCHEMA retail_pipeline;
```

```python
%sql
CREATE OR REPLACE TABLE workspace.retail_pipeline.orders AS
SELECT *
FROM VALUES
    (1, 101, 'Laptop', 1200.00, 'pending'),
    (2, 102, 'Mouse', 25.00, 'completed'),
    (3, 103, 'Keyboard', 75.00, 'completed'),
    (4, 101, 'Monitor', 300.00, 'completed'),
    (5, 104, 'Headphones', 80.00, 'pending'),
    (6, 105, 'Webcam', 60.00, 'completed'),
    (7, 102, 'USB Cable', 15.00, 'completed'),
    (8, 106, 'Desk Lamp', 45.00, 'pending')
AS t(order_id, customer_id, product, amount, status);
```

```python
%sql
CREATE OR REPLACE TABLE workspace.retail_pipeline.sales AS
SELECT *
FROM VALUES
    (1001, 2, 25.00, '2026-10-01'),
    (1002, 3, 75.00, '2026-10-01'),
    (1003, 4, 300.00, '2026-10-02'),
    (1004, 6, 60.00, '2026-10-02'),
    (1005, 7, 15.00, '2026-10-03'),
    (1006, 9, 150.00, '2026-10-03')
AS t(sale_id, order_id, amount, sale_date);
```

```python
%sql
SELECT 'orders' AS source, COUNT(*) AS rows
FROM workspace.retail_pipeline.orders

UNION ALL

SELECT 'sales', COUNT(*)
FROM workspace.retail_pipeline.sales;
```

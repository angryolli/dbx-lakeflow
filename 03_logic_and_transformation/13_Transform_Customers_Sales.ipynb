```python
from pyspark.sql.functions import (
    col, initcap, trim, year, date_format, round as spark_round
)

source = "workspace.retail_pipeline.customers_sales_silver"
target = "workspace.retail_pipeline.customers_sales_gold"

df = spark.read.table(source)

df = (
    df
    .withColumn("customer_name", initcap(trim(col("customer_name"))))
    .withColumn("customer_id", col("customer_id").cast("long"))
    .withColumn("total_price", col("total_price").cast("double"))
    .withColumn("order_year", year(col("sale_date")))
    .withColumn("order_month", date_format(col("sale_date"), "MMM"))
)

df.write.mode("overwrite").saveAsTable(target)
```

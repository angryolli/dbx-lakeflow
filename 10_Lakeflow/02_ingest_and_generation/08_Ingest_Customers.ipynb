```python
catalog = "workspace"
schema = "retail_pipeline"

source_path = (
    f"/Volumes/{catalog}/{schema}/"
    "trigger_storage_location/"
)

target_table = f"{catalog}.{schema}.customers_bronze"

customers_df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(source_path)
)

(
    customers_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_table)
)

print(f"Ingested {customers_df.count()} customers")
```
